```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
resDf = pd.read_csv("SummarizedResult.csv")
```

```python
resDf.columns
```

```python
# RMSE
Ns = resDf['N']
plt.plot(Ns, resDf['RMSE_DivQC'], label="Our method", marker="o")
plt.plot(Ns, resDf['RMSE_LowDepth'], label="Low-depth", marker="s")

plt.xlabel("N")
plt.ylabel("RMSE")
plt.xscale("log")
plt.xlim(5,100)
plt.ylim(0, 7e-4)

plt.legend()
plt.show()
```

```python
# depth
plt.plot(Ns, resDf['maxDepthAve_DivQC'], label="Our method", marker="o")
plt.plot(Ns, resDf['maxDepth_LowDepth'], label="Low-depth", marker="s")

plt.xlabel("N")
plt.ylabel("Max depth")
plt.xscale("log")
plt.yscale("log")
plt.xlim(5,100)
plt.ylim(1e3, 1e4)

plt.legend()
plt.show()
```

```python
# total query
plt.plot(Ns, resDf['totalQueryNumAve_DivQC'], label="Our method", marker="o")
plt.plot(Ns, resDf['totalQueryNum_LowDepth'], label="Low-depth", marker="s")

plt.xlabel("N")
plt.ylabel("Total query num")
plt.xscale("log")
plt.yscale("log")
NMmin, NMax = 5, 100
plt.xlim(5,100)
# plt.ylim(1e5, 1e8)

# Classical MC sample complexity
pTrue = 0.649604767419927
rmse = 0.0004
NsFine = np.linspace(NMmin, NMax, 100)
sampNumClMC = NsFine / rmse / rmse * pTrue * (1 - pTrue)
plt.plot(NsFine, sampNumClMC, label="Classical MC", linestyle="--", color='black', linewidth=0.7)

plt.legend()
plt.show()
```
