```python
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt
import datetime
from StocProcess.RBM import MakeRBMTransProbFunc
from QAE.LowDepthQAE import LowDepthQAE
```

```python
# RBM parameters
c = -1
d = 1
x0 = 0.5 * (c + d)
t0 = 0
t = 0.6
mu = 0.5
sigma = 1.0
n_terms = 100

# QAE setting
beta = 0.5
nShot = 12
epsilon = 1 / 2**15
```

```python
transProbFunc = MakeRBMTransProbFunc(t, t0, c, d, mu, sigma, n_terms)

# calculate P(x(t)>x0) as the QAE target
integFunc = lambda x: transProbFunc(x, x0)
integ, _ = sp.integrate.quad(integFunc, x0, 1.0)

# run LowDepthQAE
ldqaeResult = LowDepthQAE(integ, epsilon, nShot, beta)
print(integ, ldqaeResult.aEst)
```
